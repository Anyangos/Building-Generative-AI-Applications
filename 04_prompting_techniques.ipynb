### 1.1 Zero-Shot Prompting

### 💡 Key Prompting Paradigms Explained

To effectively build and interact with digital twins, we utilize three core prompting techniques depending on the task's complexity:

1. **Zero-Shot Prompting**
   - **What it is**: Asking the LLM to perform a task without giving it any examples of completed tasks.
   - **When to use**: For straightforward, standard tasks where the model's pre-trained knowledge is sufficient (e.g., standard text classification or simple extraction).
   - **Digital Twin Example**: *"Extract the main hobby from this text: 'I spent my weekend painting miniatures.'"*

2. **Few-Shot Prompting**
   - **What it is**: Providing the model with a few input-output examples (exemplars) to demonstrate the desired response format or style before asking for the final answer.
   - **When to use**: When the task has a highly specific output format, complex rules, or domain-specific nuances that the model might not grasp out-of-the-box.
   - **Digital Twin Example**:
     ```text
     Input: I went running for 30 minutes. -> Activity Category: Exercise
     Input: I watched a movie on the couch. -> Activity Category: Leisure
     Input: I prepared a salad for lunch. -> Activity Category:
     ```

3. **Chain-of-Thought (CoT) Prompting**
   - **What it is**: Encouraging the model to generate its intermediate reasoning steps before providing the final answer (often triggered by phrases like *"Let's think step by step"*).
   - **When to use**: For complex reasoning, logic, math, multi-step decisions, or state-tracking in user simulations.
   - **Digital Twin Example**:
     ```text
     User Profile: Works 9-5, sleeps at 10:30 PM, typically exercises after work on Tuesdays and Thursdays.
     Current State: It is 6 PM on a Thursday. The user just logged that they are feeling very exhausted from a stressful meeting.
     Question: Should the twin suggest going for a run or a relaxing evening routine? Think step-by-step.
     ```

````python
# Example prompts for digital twin
zero_shot_examples = [
    {
        'task': 'activity_prediction',
        'prompt': 'Predict the next activity: I just finished work, had dinner, and now...',
        'expected': 'leisure/relaxation activity'
    },
    {
        'task': 'preference_extraction',
        'prompt': 'Extract preferences from: I love outdoor activities but hate crowded places',
        'expected': {'likes': ['outdoor', 'quiet'], 'dislikes': ['crowds']}
    }
]

print('Zero-Shot Examples:')
for ex in zero_shot_examples:
    print(f"\nTask: {ex['task']}")
    print(f"Prompt: {ex['prompt']}")
    print(f"Expected: {ex['expected']}")
````
